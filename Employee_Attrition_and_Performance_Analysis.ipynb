```python
import pandas as pd

pd.crosstab(df['Department'], df['Attrition']).plot(kind='bar', stacked=False)
plt.title('Attrition Count by Department')
plt.xlabel('Department')
plt.ylabel('Number of Employees')
plt.legend(title='Attrition')
plt.show()
```

```python
pd.crosstab(df['OverTime'], df['Attrition']).plot(kind='bar', color=['steelblue', 'orangered'])
plt.title('Attrition by OverTime Status')
plt.xlabel('OverTime')
plt.ylabel('Number of Employees')
plt.legend(title='Attrition')
plt.show()
```

```python
plt.figure(figsize=(10,6))
pd.crosstab(df['JobRole'], df['Attrition']).plot(kind='bar', color=['steelblue', 'orangered'])
plt.title('Attrition by Job Role')
plt.xlabel('Job Role')
plt.ylabel('Number of Employees')
plt.legend(title='Attrition')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

## Part 4: Initial Business Insights

### Workforce Observations

1. The company's workforce is primarily concentrated in the Research & Development department, which accounts for approximately 65% of all employees. Sales is the second-largest department, while Human Resources represents less than 5% of the total workforce.

2. The age distribution is largely centered between 28 and 38 years, indicating that most employees are in the early to mid stages of their careers.

3. Employee salaries are heavily concentrated below $5,000 per month, with only a small proportion earning up to $20,000. This suggests the organization has a larger population of junior and mid-level employees, with relatively few senior or executive-level positions.

4. The company has an overall attrition rate of 16.1%, meaning approximately one out of every six employees has left the organization.

5. Attrition is not evenly distributed across job roles. Sales Representatives and Laboratory Technicians experience turnover rates that are noticeably higher than the company average, highlighting these roles as key areas of concern.

### Factors Contributing to Attrition

1. Overtime – Employees who regularly work overtime are significantly more likely to leave the company than those who do not, making overtime one of the strongest indicators of employee attrition in the dataset.

2. Job Role – Sales Representatives stand out with an attrition rate of nearly 50%, making them the role with the highest employee turnover across the organization.

3. Age – Employees who leave the company tend to be younger, with a median age of around 31 years, compared to a median age of approximately 36 years for employees who remain. This suggests that early-career professionals are more likely to seek opportunities elsewhere.

4. Monthly Income – Compensation appears to play an important role in employee retention. Employees who leave earn a median monthly income of about $3,000, whereas those who stay typically earn over $5,000, indicating that lower paid employees are more likely to resign.

5. Department – Although Research & Development records the highest number of employee departures due to its larger workforce, the Sales department has a higher attrition rate relative to its size. This points to retention challenges within Sales that extend beyond simply having more employees.

# Business Understanding Report

## Introduction

Employee attrition refers to the gradual reduction of a company's workforce through resignations, retirements, or terminations. For organizations such as ABC Manufacturing Ltd., understanding the reasons behind employee turnover is essential for maintaining productivity, preserving institutional knowledge, and supporting long-term business performance.

## Business Impact of Attrition

Employee turnover affects far more than the cost of replacing staff. Every departure creates additional expenses related to recruitment, onboarding, and training, while new employees require time to reach full productivity. High attrition can also lower team morale, interrupt ongoing projects, and result in the loss of valuable organizational knowledge that may have taken years to develop. In positions that depend on specialized expertise or strong client relationships, losing experienced employees can further impact customer satisfaction and business revenue.

## How Data Science Supports HR Decisions

Many organizations have traditionally relied on manager experience or exit interviews to understand why employees leave. While these approaches provide useful insights, they are often reactive because they occur after an employee has already decided to resign. Data science enables HR teams to take a more proactive approach by analyzing historical workforce data to uncover patterns and identify factors linked to employee attrition, such as overtime, compensation, job role, and tenure. These insights allow organizations to implement targeted retention strategies, including reviewing overtime policies, adjusting compensation for high-risk roles, and creating stronger career development opportunities, rather than applying broad retention initiatives across the entire workforce.

## Relevance to This Project

This project uses IBM's HR Analytics dataset to examine employee attrition patterns within a simulated manufacturing company. The objective is to uncover the key factors influencing employee turnover and generate actionable business insights. These findings will also provide a solid foundation for developing a future predictive model capable of identifying employees who may be at a higher risk of leaving, enabling HR teams to take timely and informed action.
